```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# reading data

```python
data = pd.read_csv('data/CommerceShipping.csv')
print(data.head())
```

# first lines

```python
print(data.info())
```

# Statistic info

```python
print(data.describe())
```

# Type classification

```python
categorical = data.select_dtypes(include=['object']).columns.tolist()
numerical = data.select_dtypes(include=['number']).columns.tolist()
print(f'categorical : {categorical}')
print(f'numerical : {numerical}')
```

# Mean shopping per gender

```python
print(data.groupby('Gender')['Prior_purchases'].mean())
```

# Buyers with high priority among females and and price over 200

```python
filtered_f = data[(data['Product_importance'] == 'high') &
                (data['Warehouse_block'] == 'F') &
                (data['Cost_of_the_Product'] > 200)]
print(filtered_f)
```

# Discount column

```python
data['Discount_percent'] = 100 * (data['Discount_offered'] / data['Cost_of_the_Product'])
print(data[['Discount_offered', 'Cost_of_the_Product', 'Discount_percent']].head())
```

# Mean price per block

```python
block_avg = data.groupby('Warehouse_block')['Cost_of_the_Product'].transform('mean')
data['Block_avg_cost'] = block_avg
print((data[['Warehouse_block', 'Cost_of_the_Product', 'Block_avg_cost']].head()))
```

                                        # Part 2 - merc.csv

```python
merc = pd.read_csv('data/merc.csv')
print(merc.head())
```

# Sum of cars per model

```python
model_counts = merc['model'].value_counts()
model_counts.plot(kind='bar', title='Model Counts')
plt.xlabel('Models')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

# Average price per model

```python
median_prices = merc.groupby('model')['price'].median().sort_values(ascending=False)
median_prices.plot(kind='bar', title='Median Prices')
plt.xlabel('Model')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

# Average model prices with transmission type

```python
avg_price = merc.groupby(['model', 'transmission'])['price'].mean().unstack().fillna(0)
avg_price.plot(kind='bar', title='Average Prices per model and transmission type')
plt.xlabel('Model')
plt.ylabel('Average Price')
plt.tight_layout()
plt.show()
```

# Prices histogram

```python
# overall
merc['price'].plot(kind='hist', bins=50, title='Prices Histogram')
plt.xlabel('price')
plt.show()

# per transmission type
merc.groupby('transmission')['price'].plot(kind='hist', bins=50, alpha=0.5, legend=True)
plt.title('Prices per transmission types')
plt.xlabel('Price')
plt.show()
```

# Fuel consumption diagram

```python
# overall
sns.boxplot(y=merc['mpg'], showfliers=True)
plt.title('Boxplot of fuel consumption')
plt.show()

# per fuel type
sns.boxplot(x='fuelType', y='mpg', data=merc, showfliers=True)
plt.title('Boxplot mpg of fuel consumption bases on fuel type')
plt.show()
```

# Correlation diagram

```python
sns.scatterplot(x='mileage', y='price', data=merc)
plt.title('Price vs Mileage')
plt.show()

corr_price_mileage = merc['price'].corr(merc['mileage'])
print(f'Correlation coef {corr_price_mileage:.4f}')
```

# Fuel consumption vs engine capacity

```python
sns.scatterplot(x='engineSize', y='mpg', data=merc)
plt.title('mpg vs engineSize')
plt.show()

corr_mpg_engine = merc['mpg'].corr(merc['engineSize'])
print(f"Correllation coef: {corr_mpg_engine:.4f}")
```
